```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

pd.set_option('display.max_columns',None)
```

```python
df = pd.read_csv('CC GENERAL.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=['CREDIT_LIMIT'])
df.shape
```

```python
X = df[['BALANCE','CREDIT_LIMIT']]
X.head()
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:5]
```

```python
wc = []

for i in range(2,11):
    kmeans = KMeans(
        n_clusters=i,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    wc.append(kmeans.inertia_)

plt.figure(figsize=(8,5))
plt.plot(range(2,11),wc,marker='o')
plt.xlabel('Number of Clusters')
plt.ylabel('WC')
plt.title('Elbow Method')
plt.show()
```

```python
kmeans5 = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)

clusters5 = kmeans5.fit_predict(X_scaled)
df['Cluster_5'] = clusters5
df.head()
```

```python
df['Cluster_5'].value_counts()
```

```python
centers = scaler.inverse_transform(kmeans5.cluster_centers_)
centers_df = pd.DataFrame(
    centers,
    columns=['BALANCE','CREDIT_LIMIT']
)
centers_df
```

```python
plt.figure(figsize=(10,6))

sc = plt.scatter(
    df['BALANCE'],
    df['CREDIT_LIMIT'],
    c=df['Cluster_5'],
    cmap='viridis'
)

centers_correct = scaler.inverse_transform(kmeans5.cluster_centers_)
centroids = plt.scatter(
    centers_correct[:,0],
    centers_correct[:,1],
    c='red',
    marker='X',
    s=200,
    label='Centroids'
)

handles, labels = sc.legend_elements(prop="colors")
handles.append(centroids)
labels.append('Centroids')

plt.legend(handles, labels, title="Legend")
plt.xlabel('BALANCE')
plt.ylabel('CREDIT_LIMIT')
plt.title('Credit card using k-means(k=5)')
plt.show()
```
